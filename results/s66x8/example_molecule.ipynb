```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.ticker as ticker

name = 'PyridineEthene'
path = f"./s66x8/example_molecule/{name}.csv"
data1 = pd.read_csv(path)

x1 = data1['scan']
y1 = (data1['wb97m'] - data1['wb97m'].min()) *627.5
y2 = data1['resff_relative']
y3 = data1['mace_relative']
y4 = data1['esp_relative']
y5 = data1['openff_relative']
plt.figure(figsize=(7, 4.5))

plt.plot(x1, y1, linestyle='-', color='red', label='QM', alpha=0.7, linewidth=4)
plt.plot(x1, y2, linestyle='-', color='orange', label='ResFF', alpha=0.7, linewidth=4)
plt.plot(x1, y3, linestyle='-', color='green', label='MACE', alpha=0.7, linewidth=4)
plt.plot(x1, y4, linestyle='-', color='blue', label='Espaloma-0.3', alpha=0.7, linewidth=4)
plt.plot(x1, y5, linestyle='-', color='purple', label='OpenFF-2.1.0', alpha=0.7, linewidth=4)
plt.ylim(-0.2,6)
plt.xlabel(r'$r/r_0$', fontsize=16)
plt.ylabel('ΔE (kcal/mol)', fontsize=16)
plt.legend(fontsize=14, frameon=False) 
plt.gca().yaxis.set_major_locator(ticker.MultipleLocator(2))
plt.tick_params(axis='both', labelsize=14)
#plt.savefig(f"./s66x8/example_molecule/{name}.png", dpi=600)  

plt.show()

```
