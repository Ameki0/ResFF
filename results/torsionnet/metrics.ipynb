```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Load Data 
file_path_resff = "resff-torsionnet.csv"
df_resff = pd.read_csv(file_path_resff)

file_path_mace = "mace-torsionnet.csv"
df_mace = pd.read_csv(file_path_mace)

file_path_gaff = "gaff-torsionnet.csv"
df_gaff = pd.read_csv(file_path_gaff)

file_path_esp = "espaloma-torsionnet.csv"
df_esp = pd.read_csv(file_path_esp)

file_path_open = "openff-torsionnet.csv"
df_openff = pd.read_csv(file_path_open)

methods = {
    "ResFF": (df_resff, "wb97m_relative_energy", "resff_relative_energy"),
    "MACE": (df_mace, "wb97m_relative_energy", "MACE_relative_energy"),
    "GAFF": (df_gaff, "b3lyp_relative_energy", "mm_relative_energy"),
    "Espaloma": (df_esp, "b3lyp_relative_energy", "mm_relative_energy"),
    "OpenFF": (df_openff, "b3lyp_relative_energy", "mm_relative_energy")
}

results = []

# Compute Metrics for All Methods
for name, (df, ref_col, pred_col) in methods.items():

    clean_df = df[[ref_col, pred_col]].dropna()
    
    y_true = clean_df[ref_col]
    y_pred = clean_df[pred_col]
    
    # Calculate RMSE, MAE, and Pearson Correlation (r)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)
    correlation = np.corrcoef(y_true, y_pred)[0, 1]
    r2 = r2_score(y_true, y_pred)

    results.append({
        "Method": name,
        "RMSE (kcal/mol)": rmse,
        "MAE (kcal/mol)": mae,
        "Correlation (r)": correlation,
        'R² Score': r2
    })

df_summary = pd.DataFrame(results)

print("\n--- Summary of TorsionNet Benchmarks ---")
print(df_summary.to_string(index=False, float_format=lambda x: f"{x:.4f}"))

```
